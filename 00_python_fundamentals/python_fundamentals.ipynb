## 3. Variables — giving a value a name

A **variable** is a **name** that refers to a **value**.

- **Assigning** means: “remember this value under this name”
- The symbol `=` in Python means **assign**, not “equals” in the maths sense
- After you assign, you can use the name anywhere you would use the value

### See assignment in action

1. Print a piece of text directly  
2. Assign that **same** text to a variable  
3. Print the variable (no quotes around the name)  
4. Print `"currency"` *with* quotes — notice this is different  

You should see `USDT` twice from steps 1 and 3. Step 4 prints the word `currency` itself, **not** the value stored in the variable.


````python
# 1) Print a string directly (a "literal" — the value itself)
print("USDT")

# 2) Assign that same string to a name (a variable)
currency = "USDT"

# 3) Print the variable — no quotes: Python looks up the name and shows its value
print(currency)

# 4) Quotes around the name → this is just text spelling "currency", NOT the variable
print("currency")
````

### What just happened?

| Line | Meaning |
|------|---------|
| `print("USDT")` | Show the text `USDT` |
| `currency = "USDT"` | Create a name `currency` and store `"USDT"` in it |
| `print(currency)` | Show whatever `currency` currently holds (here: `USDT`) |
| `print("currency")` | Show the text `currency` — quotes mean “this exact text”, not the variable |

**Rule of thumb:** quotes → literal text. No quotes → use the variable’s value.

You can **reassign** a variable later — the name then refers to a new value:


````python
currency = "USDT"
print(currency)

currency = "BTC"   # same name, new value
print(currency)
````

## 5. Arithmetic

Once you have numbers (`int` or `float`), you can calculate with them.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | Add | `1000 + 250` | `1250` |
| `-` | Subtract | `1000 - 250` | `750` |
| `*` | Multiply | `2 * 70000` | `140000` |
| `/` | Divide | `10 / 4` | `2.5` (always a float) |
| `//` | Divide, whole number only | `10 // 4` | `2` |
| `%` | Remainder after division | `10 % 4` | `2` |
| `**` | Power (to the power of) | `2 ** 3` | `8` |

Parentheses `( )` control order, same idea as maths: things inside brackets happen first.


````python
amount = 1000.0
fee = 1.5
price = 70000.0
btc_amount = 2.0

print(amount + fee)           # add
print(amount - fee)           # subtract
print(btc_amount * price)     # multiply → USD value
print(amount / 4)             # divide
print(2 ** 3)                 # 2 to the power of 3

# Parentheses change the order
print(10 + 5 * 2)             # 20  (multiply first)
print((10 + 5) * 2)           # 30  (brackets first)
````

### Updating a variable with arithmetic

A common pattern: read the current value, calculate, store the result back in the same name (you'll use this when totalling amounts in a loop).


````python
total = 0.0
total = total + 1000.0   # add first deposit
total = total + 250.5    # add second deposit
print(total)

# Shorthand (same meaning): total += 40.0
total = total + 40.0
total += 40.0
print(total)
````

## 6. Lists

A **list** is an ordered collection of items inside square brackets `[...]`, separated by commas.

- `amounts[0]` — first item (again, counting starts at 0)  
- `len(amounts)` — how many items  
- `amounts.append(...)` — add one more item on the end

### Looping with `for`

A `for` loop repeats the indented lines **once for each item** in the list.

Read `for amount in amounts:` as: “for each value in `amounts`, temporarily call it `amount`, then run the indented block.”


````python
amounts = [5.99997, 349999.0, 8333.33]

total = 0.0
for amount in amounts:      # each time around, amount is the next list item
    print(amount)           # show this item
    total = total + amount  # add it to our running total

print(f"Total: {total}")
````

Below we can show this in action, by seeing the total being increased in real time

````python
deposit_amounts = [1000.0, 250.5, 40.0, 9999.99]

# Your code here
total = 0.0
for amount in deposit_amounts:
    print(f'Current total: {total}')
    print(f'amount: {amount}')
    total = total + amount
    print(f'New total: {total}')
    print()

print(f"Total deposits: {total}")
````

## 8. Sets — unique values

A **set** is a collection of values with **no duplicates** and **no fixed order**.

Written with curly braces like a dictionary, but with **values only** (no `key: value` pairs):

```python
{"BTC", "USDT", "USDC"}
```

The most useful trick for beginners: turn a **list** into a **set** to keep only the unique items — e.g. unique tickers or networks from a list of transfers.


````python
tickers = ["BTC", "USDT", "BTC", "USDC", "USDT", "BTC"]
print(tickers)
print(len(tickers))  # 6 items, including repeats

unique_tickers = set(tickers)  # cast list → set (duplicates removed)
print(unique_tickers)
print(len(unique_tickers))  # 3 unique values

# Need a list again later? Cast back:
unique_list = list(unique_tickers)
print(unique_list)
````

## 9. Operators and boolean values

### Comparison operators

These ask a yes/no question about two values. The answer is always a **boolean**: `True` or `False`.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `amount == 1000` | `True` if equal |
| `!=` | Not equal to | `currency != "BTC"` | `True` if different |
| `>` | Greater than | `amount > 5000` | `True` if larger |
| `<` | Less than | `amount < 100` | `True` if smaller |
| `>=` | Greater than or equal | `amount >= 1000` | … |
| `<=` | Less than or equal | `amount <= 1000` | … |

**Important:** `==` means “is equal to”. A single `=` means **assign** (store in a variable). Mixing them up is a very common beginner mistake.

### Comparisons resolve to booleans

You can `print` a comparison on its own — Python evaluates it to `True` or `False`. An `if` statement uses that same idea: it runs the indented block only when the condition is `True`.


````python
amount = 8333.33
currency = "USDT"

print(amount > 5000)          # True
print(amount == 1000)         # False
print(currency == "USDT")     # True
print(currency != "BTC")      # True

# Store the result in a variable if you want
is_large = amount > 10000
print(is_large)
print(type(is_large))         # <class 'bool'>
````

### Combining conditions: `and`, `or`, `not`

| Operator | Meaning |
|----------|---------|
| `and` | Both sides must be `True` |
| `or` | At least one side must be `True` |
| `not` | Flips `True` ↔ `False` |

These also produce a boolean.

**Note:** In Python we write `and` / `or` (words), not `&` / `|`, for these yes/no conditions.


````python
transfer = {"type": "withdrawal", "amount": 15000.0, "currency": "USDT"}

print((transfer["type"] == "withdrawal") and (transfer["amount"] > 5000))  # True
print((transfer["currency"] == "BTC") or (transfer["currency"] == "USDT"))  # True
print((transfer["currency"] == "BTC") or (transfer["currency"] == "ETH"))  # False
print(not transfer["type"] == "deposit")  # True (because it is a withdrawal)
````

### How a combined condition resolves (step by step)

Python does not jump straight to the final answer. It:

1. Resolves each small comparison to `True` or `False`
2. Combines those booleans with `and` / `or` / `not`
3. Ends up with one final boolean

**Example A — all true → `True`**

```text
(10 > 5) and (9 == 9) and (len("abc") == len("def"))
→  True  and  True  and  True
→  True
```

(`len("abc")` and `len("def")` are both `3`, so they are equal.)

**Example B — one false with `and` → `False`**

```text
(10 > 5) and (9 == 8) and (len("abc") == 3)
→  True  and  False  and  True
→  False
```

(`and` needs *every* part to be `True`.)

**Example C — `or` only needs one true → `True`**

```text
(currency == "BTC") or (currency == "USDT") or (amount < 0)
→  False  or  True  or  False     (if currency is "USDT" and amount is positive)
→  True
```

**Example D — `not` flips the result**

```text
not (amount > 10000)
→  not True      (if amount is e.g. 15000)
→  False

not (amount > 10000)
→  not False     (if amount is e.g. 500)
→  True
```

Run the cell below to see Python print the same step-by-step idea.


````python
# Example A — all parts True
print("Example A")
print(10 > 5)                    # True
print(9 == 9)                    # True
print(len("abc") == len("def"))  # True (both lengths are 3)
print((10 > 5) and (9 == 9) and (len("abc") == len("def")))  # True

# Example B — and with one False
print("\nExample B")
print(10 > 5)           # True
print(9 == 8)           # False
print(len("abc") == 3)  # True
print((10 > 5) and (9 == 8) and (len("abc") == 3))  # False

# Example C — or
print("\nExample C")
currency = "USDT"
amount = 500.0
print(currency == "BTC")    # False
print(currency == "USDT")   # True
print(amount < 0)           # False
print((currency == "BTC") or (currency == "USDT") or (amount < 0))  # True

# Example D — not
print("\nExample D")
amount = 15000.0
print(amount > 10000)           # True
print(not (amount > 10000))     # False

amount = 500.0
print(amount > 10000)           # False
print(not (amount > 10000))     # True
````

## 10. Conditionals (`if` / `elif` / `else`)

Sometimes you only want code to run **when a condition is true**.

A condition is any expression that resolves to a boolean (`True` or `False`) — the comparisons and `and` / `or` / `not` from the previous section.

- `if ...:` — run the indented block when the condition is `True`  
- `elif ...:` — (“else if”) another condition to try if the previous ones were `False`  
- `else:` — run if none of the conditions above were `True`  

Indentation (the spaces at the start of the line) matters: it shows Python which lines belong to the `if`.


````python
transfer = {"type": "withdrawal", "amount": 12000.0, "currency": "USDT"}

if transfer["type"] == "deposit":
    print("Incoming funds")
elif transfer["amount"] > 10000:
    print("Large withdrawal")
else:
    print("Regular withdrawal")
````

### What `if` is really checking

Remember: a comparison like `transfer["type"] == "deposit"` evaluates to `True` or `False`.

So this:

```python
if transfer["type"] == "deposit":
```

is the same idea as:

```python
if (transfer["type"] == "deposit") == True:
```

You can also store the comparison in a variable first, then use that variable in the `if`:

```python
is_deposit = transfer["type"] == "deposit"  # is_deposit is True or False
if is_deposit:
```

All three styles check the same thing. You almost never write the `== True` version — it's redundant — but seeing all three helps show *why* `if` works: it only runs the indented block when the condition is the boolean `True`.


````python
transfer = {"type": "deposit", "amount": 1000.0, "currency": "USDT"}

# The comparison itself is True or False
print(transfer["type"] == "deposit")

# Store that boolean in a variable
is_deposit = transfer["type"] == "deposit"
print(is_deposit)

# These three ifs do the same thing
if transfer["type"] == "deposit":
    print("Incoming funds (normal style)")

if (transfer["type"] == "deposit") == True:
    print("Incoming funds (explicit == True)")

if is_deposit:
    print("Incoming funds (using a boolean variable)")
````

Filter a list of transfers — the same idea you'll later do with pandas, but by hand first.

## 11. Functions

A **function** is a reusable recipe: you **define** it once, then **call** it whenever you need it.

- `def name(inputs):` starts the definition  
- Indented lines are the steps  
- `return` sends a result back to whoever called the function  

Example: `usd_value(2.0, 70000)` means “run `usd_value` with `amount=2.0` and `price=70000`”.


````python
def usd_value(amount, price):
    return amount * price

btc_usd = usd_value(2.0, 70000)
eth_usd = usd_value(100, 4000)
print(eth_usd)
````

````python
def is_large_transfer(transfer, threshold):
    return transfer["amount"] > threshold

sample = {"type": "withdrawal", "amount": 15000.0, "currency": "USDT"}
print(is_large_transfer(sample, 10000))
print(is_large_transfer(sample, 20000))
````
